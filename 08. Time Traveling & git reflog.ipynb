# 08. Time Traveling & `git reflog`

## 1. What is `git reflog`?

* **`git reflog`** (Reference Log) is an advanced companion to `git log --oneline` that tracks **every movement of `HEAD`**—including commits, checkouts, and resets.
* It is essential for **time traveling** (undoing mistakes, recovering accidentally deleted files, or jumping back to an earlier state).

```bash
git reflog
```

* **What it shows:**
  * Short **Commit ID** (first 6–7 characters)
  * **`HEAD@{0}`** (current position), **`HEAD@{1}`** (1 step back), **`HEAD@{2}`** (2 steps back), etc.
  * Action/commit message
* *Tip:* If you have a long log and the terminal pauses, press **`q`** to quit back to the prompt.

## 2. Time Traveling Back with `git reset --hard`

Suppose you accidentally deleted a file (e.g., `data_ingest.txt`) and committed the deletion. You can bring it back by resetting `HEAD` to the commit right before the deletion.

### Option A: Using Commit ID (Recommended)
Copy the short commit ID from `git reflog` and reset directly to it:

```bash
git reset --hard <commit_id>
```

* **Why Commit ID is recommended:** `HEAD@{n}` indexes shift dynamically every time you move or reset, whereas a **Commit ID** is the permanent source of truth.

### Option B: Using Relative `HEAD` (`HEAD~1`)
Jump back 1 commit from your current `HEAD`:

```bash
git reset --hard HEAD~1
```

* *Note:* Use `HEAD~2` to go back 2 commits, `HEAD~3` for 3 commits, etc.

## 3. Jumping Forward or Back Again

* Even after running `git reset --hard`, your previous commits are still recorded in `git reflog`.
* You can run `git reflog` again, grab the commit ID of the state you moved away from, and run `git reset --hard <commit_id>` to jump right back to it.
