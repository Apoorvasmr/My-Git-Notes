# 12. GitHub & Remote Workflows

## 1. Connecting Local Git to Your GitHub Account

Make sure your global Git config matches your GitHub account email and username:

```bash
git config --global user.email "your_github_email@gmail.com"
git config --global user.name "your_github_username"
```

### Creating a Personal Access Token (PAT)
To push code via **HTTPS**, GitHub requires authentication (either browser sign-in or a **Personal Access Token**):
1. On GitHub, click your **Profile Picture** -> **Settings**.
2. Scroll down to **Developer settings** -> **Personal access tokens** -> **Tokens (classic)**.
3. Click **Generate new token (classic)**, set permissions, and copy the token immediately.
4. Store it securely (you will not be able to view it again). Once authenticated, Git caches it so you don't have to enter it every time.

## 2. Scenario 1: Pushing an Existing Local Repo to GitHub

Use this workflow when you already created a Git repository locally and want to host it on a new GitHub repository.

### Step 1: Link Your Local Repo to Remote (`origin`)
```bash
# Check existing remotes (returns nothing initially)
git remote -v

# Add the GitHub repo HTTPS URL as "origin"
git remote add origin "[https://github.com/](https://github.com/)<username>/<repo_name>.git"

# Verify origin is added
git remote -v
```

### Step 2: Rename `master` to `main` (Optional Best Practice)
GitHub uses `main` as the default branch name:
```bash
git switch master
git branch -m main
```

### Step 3: Pull Unrelated Histories & Push
If your new GitHub repo was created with a `README.md` (an initial commit that doesn't exist locally), `git push origin main` will be rejected. Pull and merge the unrelated histories first:

```bash
# Pull remote main and allow merging unrelated histories
git pull origin main --allow-unrelated-histories

# Push your local main branch to GitHub
git push origin main
```

## 3. Scenario 2: Cloning a Remote Repo (Standard Industry Workflow)

In most real-world projects, the repository already exists on GitHub. You **clone** it locally instead of running `git init`.

```bash
# Clone the repository (automatically sets up "origin" for you)
git clone [https://github.com/](https://github.com/)<username>/<repo_name>.git

# Move into the cloned project folder
cd <repo_name>

# Verify remote origin is already configured
git remote -v
```

### Working on a Feature Branch
Never commit directly to `main` after cloning. Always create a feature branch:

```bash
# Create and switch to a new feature branch
git switch -c feature_1

# Make your changes, then stage and commit
git add .
git commit -m "add feature 1"

# Push the feature branch to GitHub
git push origin feature_1
```

## 4. Merging via Pull Requests (PR) on GitHub

> **Industry Best Practice:** Instead of merging your feature branch into `main` locally, push your feature branch to GitHub and merge it through a **Pull Request (PR)** so teammates can review your code.

1. Open your repository on GitHub—you will see a banner for your recently pushed branch (`feature_1`).
2. Click **Compare & pull request**.
3. Add a clear **Title** and **Description**, and review the file changes (using **Split** or **Unified** view).
4. Click **Create pull request**.
5. Once reviewed/approved, click **Merge pull request** -> **Confirm merge**.
6. Optionally click **Delete branch** to keep the repository clean.
