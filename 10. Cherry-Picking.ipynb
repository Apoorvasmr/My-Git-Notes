# 10. Git Cherry-Pick

## 1. What is Cherry-Picking?

* **Cherry-picking** allows you to pick **specific individual commits** from one branch and apply them directly onto another branch (such as `master`/`main`), without merging or rebasing the entire branch.
* **Merge vs. Cherry-Pick:**
  * **`git merge` / `git rebase`**: Brings over *all* commits from a feature branch.
  * **`git cherry-pick`**: Brings over *only* the exact commit(s) you select.
* **Important Detail:** When a commit is cherry-picked onto the target branch, Git creates a copy of those changes and assigns it a **brand-new Commit ID** (different from the original commit ID).

## 2. Best Real-World Use Case

* **Urgent Production Hotfixes:** If you are working in a `dev` or feature branch that has both an urgent bug fix and unfinished development work, you can cherry-pick **only the bug-fix commit** directly into `production`/`master` without promoting incomplete features.

## 3. Step-by-Step Cherry-Pick Workflow

### Step 1: Find the Commit ID on the Source Branch
```bash
# Switch to the branch that has the commit you want
git switch rebase_feat

# List commits and copy the short Commit ID you need
git log --oneline
```

### Step 2: Switch to the Target Branch and Cherry-Pick
```bash
# Switch to the branch where you want to apply the commit
git switch master

# Apply the specific commit
git cherry-pick <commit_id>
```

## 4. Cherry-Picking Multiple Commits

You can cherry-pick multiple commits in a single command:

```bash
git cherry-pick <commit_id_1> <commit_id_2>
```

* **Order Matters:** Always list the commit IDs in the **exact chronological order** they were originally created (oldest first). Applying them out of order—especially when they modify the same file—can cause merge conflicts.
