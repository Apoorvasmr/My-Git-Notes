# 02. Git Configuration & Repository Initialization

## 1. Opening the Terminal in VS Code

* **Shortcut:** Press `Ctrl + Shift + ~` (backtick) or go to `Terminal -> New Terminal`.
* **PowerShell vs. Command Prompt:** Prefer **PowerShell** (or **Git Bash**) over Windows Command Prompt, as Command Prompt does not support common Linux commands like `ls` or `touch` that are frequently used alongside Git.

## 2. Checking Status Before Initializing

* Always make it a habit to run `git status` first when working in a folder:

```bash
git status
If the folder is just a normal directory and not yet tracked by Git, you will see:fatal: not a git repository (or any of the parent directories): .git

3. Global Git Configuration
Before creating commits, tell Git who you are. This identity is attached to every commit you make and is used when pushing code to remote platforms like GitHub.

Bash
git config --global user.email "your_email@gmail.com"
git config --global user.name "your_username"
Why --global? Saves your credentials system-wide so you do not have to re-enter them for every new repository.

Best Practice: Use the same email and username as your GitHub account.

4. Initializing a Git Repository
Convert a normal folder into a Git repository using:

Bash
git init
What happens behind the scenes?

Git creates a hidden .git folder inside your project directory.

The .git folder is the brain/database of your repository—it stores all commit history, versions, and metadata.

Warning: Do not manually edit or delete files inside .git.

Viewing .git in VS Code (Optional): Go to Settings -> search files.exclude -> remove **/.git.

5. Registering the Default Branch (Initial Commit)
Right after git init, running git status shows On branch master (or main), but running git branch returns nothing.

Why? Git does not officially register a branch until it has at least one commit.

Best Practice: Create a README.md (Markdown) file to introduce the repository and make your first commit:

Bash
# Stage all files in the directory
git add .

# Create the initial commit
git commit -m "initial commit"
Verify:

Run git branch — it will now display master (or main).
