# 06. Merge Conflicts

## 1. Why Do Merge Conflicts Happen?

* A **merge conflict** occurs when two different branches (e.g., `master` and `bug_fix`) start from the same **common ancestor (`v0`)** and both modify the **same file** independently.
* When you try to merge them, Git sees two competing versions (`v1` from `master` and `v1` from `bug_fix`) and pauses the merge so **you** can decide which changes to keep.
* *Note:* Even if two developers edit different parts of the same file, Git can still flag a conflict if the modified regions overlap or clash with the ancestor state.

## 2. Aborting a Merge

If you run into a conflict and want to cancel the merge completely and return to your pre-merge state:

```bash
git merge --abort
```

* Verify everything is clean afterward with `git status`.

## 3. Resolving Merge Conflicts in VS Code

When `git merge <branch_name>` triggers a `CONFLICT`, open the conflicted file in VS Code. You have two ways to resolve it:

### Option A: Using VS Code Merge Editor (Recommended)
1. Click **Resolve in Merge Editor** at the bottom right of the conflicted file.
2. You will see three panes:
   * **Incoming (Left):** Changes coming from the feature/bug-fix branch.
   * **Current (Right):** Changes currently on your target branch (e.g., `master`).
   * **Result (Bottom):** The final output file.
3. Choose **Accept Current**, **Accept Incoming**, or **Accept Combination (Both)**.
4. Click **Complete Merge**.

### Option B: Resolving Manually (Inline)
1. Open the file directly—you will see inline conflict markers (`<<<<<<<`, `=======`, `>>>>>>>`) separating the Current and Incoming changes.
2. Either click the quick-action links above the block (**Accept Current Change** / **Accept Incoming Change** / **Accept Both Changes**) **or** manually delete all conflict markers and edit the file to the exact final version you want.
3. Save the file (`Ctrl + S`).

## 4. Finalizing the Merge (Non-Negotiable)

Resolving the file alone does **not** finish the merge. If you run `git status`, Git will tell you:
`All conflicts fixed but you are still merging.`

You **must** stage and commit the resolved file to conclude the merge:

```bash
# Step 1: Stage the resolved file(s)
git add .

# Step 2: Create the merge commit
git commit -m "resolve merge conflict"
```

* Verify the completed merge in **Git Graph** or with `git status`.
