# 11. Git Stashing

## 1. What is Stashing?

* **Stashing** is a **temporary storage location** provided by Git where you can safely shelve uncommitted changes from your Working Directory.
* **Real-World Scenario:**
  * You are mid-way through modifying a tracked file on a feature branch (`dev_1`), and your code is **not** ready to commit yet.
  * Suddenly, you need to switch to `master` to fix an urgent bug.
  * If you run `git switch master`, Git blocks you with:
    `error: Your local changes to the following files would be overwritten by checkout... Please commit your changes or stash them.`
* **Solution:** Stash your in-progress work, switch branches to fix the bug, and restore your stashed work when you return.

## 2. Saving Changes to Stash

```bash
# Save modified changes to stash with a meaningful message (Recommended)
git stash push -m "my development paused"
```

* Your Working Directory is immediately cleaned (changes are moved to temporary storage), allowing you to freely run `git switch master`.

## 3. Viewing Your Stashes (`git stash list`)

```bash
git stash list
```

* **Stacked Memory (LIFO):** Stashes work like a stack—the **latest stash** is always at the top as `stash@{0}`, followed by older ones (`stash@{1}`, `stash@{2}`, etc.), along with the branch name and message.
* *Note:* Stashes are accessible across the entire repository, so they can be viewed or applied from any branch.
