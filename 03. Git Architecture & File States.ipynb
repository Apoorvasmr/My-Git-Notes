# 03. Git Architecture, File States & HEAD

## 1. The `.git` Folder (The Brain)

* The **`.git`** folder is the **database / backbone** of your Git repository.
* It stores all versions, commit history, checkpoints, and metadata.
* Without `.git`, a folder is just a normal directory. With `.git`, it becomes a **Repository** capable of version control and time-traveling.

## 2. Git Workflow: The 3 Areas

Every change in Git moves through three distinct stages:

**`Working Directory` $\rightarrow$ `Staging Area` $\rightarrow$ `Repository (.git)`**

1. **Working Directory:**
   * Your active project folder where you create, edit, and modify files.
   * Git does not automatically track new files or new edits made here until you explicitly tell it to.
2. **Staging Area (Snapshot Area):**
   * A middle ground where you bundle specific files/changes together to take a **snapshot** before saving them permanently.
   * **Copy-Paste Analogy:** Selecting specific files in File Explorer (`Ctrl + C`) is **Staging**; pasting them to save (`Ctrl + V`) is **Committing**.
3. **Repository (`.git`):**
   * The permanent database where committed snapshots are stored as **Commits** (checkpoints).

## 3. File States in Git

Files in a Git project exist in one of four states (VS Code shows single-letter badges next to filenames):

| State | VS Code Badge | Meaning |
| :--- | :---: | :--- |
| **Untracked** | `U` | Brand-new file in the Working Directory; Git sees it exists but is not tracking it yet. |
| **Modified** | `M` | An already-tracked file has been edited, but the new changes are not staged yet. |
| **Staged** | `A` | Changes have been added to the Staging Area (snapshot ready to be committed). |
| **Committed** | *(none)* | Snapshot is safely stored in the `.git` repository. |

## 4. Hands-On Commands: Creating, Staging & Committing

### Creating a File from the Terminal
```bash
echo "This is another data ingestion file" > data_ingest.txt
```

### Staging Files (`git add`)
```bash
# Stage a single specific file
git add data_ingestion.txt

# Stage multiple specific files
git add data_ingest.txt iot_data_ingestion.txt

# Stage ALL new and modified files at once
git add .
```
* *Note:* `git add .` only stages files that are new or modified—unchanged files are skipped.

### Committing Changes (`git commit`)
```bash
# Commit staged changes with a message
git commit -m "ingest data"
```
* **Commit Message Best Practices:**
  * Keep messages **brief** (2–5 words / a short phrase).
  * Write in **simple present tense** (imperative style), as if giving Git a command (e.g., `"ingest data"` instead of `"data ingested"`).
* **What if you forget `-m`?**
  * Running `git commit` without `-m` opens your default editor (**VS Code**) and waits for you to type a commit message, save (`Ctrl + S`), and **close the tab** to complete the commit.
* **Bundling:** Multiple files staged together before a single `git commit` are recorded as **one commit** in Git Graph.

## 5. What is `HEAD`?

* **`HEAD`** is a pointer that always points to the **latest commit** on your **currently active branch** (e.g., `master` or `main`).
* You can see it in **Git Graph** or inside `.git/HEAD` (which shows `ref: refs/heads/master`).
* **Why `HEAD` matters:** Whenever you create a new branch, Git creates it from the **`HEAD`** of your current branch by default.
