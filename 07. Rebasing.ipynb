# 07. Git Rebase

## 1. What is Rebasing?

* **Rebasing** means **shifting the base (origin point)** of your feature branch from an older commit to the latest commit (`HEAD`) of another branch (typically `master`/`main`).
* **Merge vs. Rebase:**
  * A normal **Three-Way Merge** keeps both divergent histories and ties them together with an extra **Merge Commit**.
  * **Rebase** replays your feature branch commits on top of the latest `master` commit, creating a clean, straight-line history.

## 2. Why Use Rebase? (Rebase + Fast-Forward Merge)

* After rebasing your feature branch onto the latest `master`, `master` has no new commits ahead of your branch's base.
* Merging the feature branch into `master` then becomes a **Fast-Forward (`FF`) Merge**—no extra merge commit, no messy side branches, and a clean, linear commit history in **Git Graph**.

## 3. Step-by-Step Rebase Workflow

```bash
# Step 1: Switch to the feature branch you want to rebase
git switch rebase_feat

# Step 2: Shift its base to the latest commit (HEAD) of master
git rebase master

# Step 3: Switch back to master
git switch master

# Step 4: Perform a clean Fast-Forward merge
git merge rebase_feat
```
## 4. Handling Rebase Conflicts

* If a commit on your feature branch modifies the same file as a newer commit on `master`, Git will pause the rebase and flag a conflict.
* Resolve it the **exact same way** as a merge conflict (via VS Code Merge Editor or manually), then stage the file and continue.
