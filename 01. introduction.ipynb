# 01. Introduction & Environment Setup

## 1. Environment Setup

* **Code Editor (VS Code):** Widely adopted across Windows, macOS, and Linux. It allows you to manage project files and run Git terminal commands in one place.
* **Why start with plain `.txt` files?** Learning Git with simple text files keeps the focus entirely on Git concepts and tracking behavior rather than programming language syntax or interpreters.
* **Installing Git:** Download the OS-compatible installer from the official Git downloads page.
  * **Git Bash:** Check the option to add the Git Bash profile to Windows Terminal. It supports Linux-style commands (like `ls` and `touch`) that standard Windows Command Prompt does not, though **PowerShell** inside VS Code works well too.
  * **Default Editor:** Change the default editor from **Vim** to **Visual Studio Code** (`Use Visual Studio Code as Git's default editor`). This prevents getting stuck in Vim if you forget to include a commit message.
* **Workspace Setup:** Create a dedicated empty folder (e.g., `git_practice`) and open it in VS Code via `File` -> `Open Folder`.
* **Recommended VS Code Extension:** Install **Git Graph** from the Extensions marketplace to visually track your commits, branches, and progress.

## 2. Core Concepts

### What is Git?
* Git is a **Version Control System (VCS)** — essentially a **"time machine for your code."**
* **Video Game Analogy:** Just like saving a checkpoint in a game so you can resume later or reload if you fail a mission, Git lets you save code checkpoints called **Commits**.
* **Key Capabilities:**
  * Track every change made to your code over time.
  * Travel back to older versions if new changes break something.
  * Collaborate with teammates in isolated environments without overwriting each other's work.
* **Why Git became the standard:** Launched around 2005, it was **open-source** (free) and solved version-control problems that expensive paid tools struggled with.
* **Under the hood:** Git is the foundational software. Managed platforms like **GitHub** and **Azure DevOps** are built on top of Git.

### What is GitHub?
* A **cloud-based managed platform** used to host Git repositories online.
* Enables teams to store code remotely (in **public** or **private** repos), collaborate, review code, and track issues.

| Feature | Git | GitHub |
| :--- | :--- | :--- |
| **Type** | Version control software / tool | Cloud-based hosting platform |
| **Where it runs** | Locally on your computer | Remotely on the internet (cloud) |
| **Primary Purpose** | Track changes, create commits, time-travel | Store, share, collaborate, review code |

### What is a Repository (Repo)?
* **Repo** is short for **Repository** — simply a **root/parent project folder tracked by Git** (`Normal Folder + Git Tracking = Repo`).
* **Contents:** Holds your code, documentation, and the complete history of changes.
* **Storage:** Can exist **locally** (on your machine) or **remotely** (on GitHub).
* **Important Note:** Creating a normal folder (directory) does *not* automatically make it a Git repository — it must be initialized with Git first.
