# 05. Branches & Merging

## 1. What Are Branches?

* Branches are the **heart of Git**—they allow multiple developers to work on the same project **in parallel** without interfering with each other's code.
* **Parent Branch (`master` / `main`):** The default main branch of your repository (`main` is the modern name; `master` is the older default—both mean the same thing).
* **Feature Branch:** An **isolated environment** created from the parent branch to build a specific feature or fix a bug.
* **Ancestor Commit:** By default, a new branch is created from the **`HEAD`** (latest commit) of your currently active branch and inherits all of its previous commits.
* **Isolation:** Changes and commits made on a feature branch are completely invisible to `master` (and other branches) until they are merged.

> **Best Practice:** Always switch to `master`/`main` **first** before creating a new feature branch, so you don't accidentally branch off another feature branch.

## 2. Branching Commands

### List All Branches
```bash
git branch
```
* Displays all local branches. The **`*`** (highlighted in green) marks the branch you are currently on.

### Switch & Create Branches (`git switch` — Modern, Recommended)
Introduced in 2019 specifically for branch operations:

```bash
# Switch to an existing branch
git switch master

# Create a new branch AND switch to it immediately (-c = create)
git switch -c feature_1
```

### Switch & Create Branches (`git checkout` — Older Alternative)
An older, multi-purpose command still widely seen in teams:

```bash
# Switch to an existing branch
git checkout master

# Create a new branch AND switch to it immediately
git checkout -b feature_1
```

## 3. Merging Branches (`git merge`)

Once work on a feature branch is complete, bring those commits into `master`/`main`:

```bash
# Step 1: Switch to the target branch (where you want to bring the changes)
git switch master

# Step 2: Merge the feature branch into master
git merge feature_1 -m "merge feature_1"
```

* *Note:* If you omit `-m` during a non-fast-forward merge, Git opens VS Code with a default merge message—save and close the tab to complete the merge.
* *Tip:* You can verify the merged history visually in **Git Graph** or via `git log --oneline`.

## 4. Types of Merges

| Merge Type | When It Happens | Result |
| :--- | :--- | :--- |
| **Fast-Forward Merge** | `master` has **no new commits** since the feature branch was created. | Git simply moves the `master` pointer forward to the feature branch's latest commit. **No extra merge commit** is created. |
| **Non-Fast-Forward (Three-Way) Merge** | **Both** `master` and the feature branch have new commits after diverging. | Git combines both histories and creates a brand-new **Merge Commit**. |
