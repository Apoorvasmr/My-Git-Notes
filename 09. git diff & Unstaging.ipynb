# 09. `git diff` & Unstaging Changes

## 1. What is `git diff`?

* **`git diff`** shows the exact line-by-line differences between two states—such as your Working Directory vs. Staging Area, Staging Area vs. Repository, or two different branches/files.

### Compare Working Directory vs. Staging Area
Shows changes you have made in your files that have **not** been staged yet:

```bash
git diff
```

### Compare Staging Area vs. Repository (Last Commit)
Once you have run `git add .`, use `--staged` to see what is about to be committed:

```bash
git diff --staged
```

## 2. How to Read `git diff` Output

* **`--- a/<file>`**: Source / original version of the file.
* **`+++ b/<file>`**: Destination / updated version of the file.
* **`@@ ... @@`**: Line number indicators showing where the changes occur.
* **`+`** (green): Lines added.
* **`-`** (red): Lines removed.

## 3. Unstaging Changes (`git reset`)

If you staged files with `git add .` by mistake and want to pull them back out of the Staging Area **without** losing your edits:

```bash
# Unstage ALL staged files (moves them back to Modified in Working Directory)
git reset

# Unstage only a specific file
git reset <file_name>
```

* **What happens:** Your file moves from **Staged (`A`)** back to **Modified (`M`)** in the Working Directory—no commit ID is needed.
* **Discarding Working Directory Changes:** If you also want to erase the edits completely, open the **Source Control** panel in VS Code and click **Discard Changes** on the file.
