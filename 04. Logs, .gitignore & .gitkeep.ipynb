# 04. Logs, `.gitignore` & `.gitkeep`

## 1. Viewing Commit History (`git log`)

Logs store the tracking metadata and history of all commits in your repository. You need logs whenever you want to inspect past changes or time-travel to an older commit.

### Full Commit Log
```bash
git log
```
* Displays the full commit history: full commit hash ID, author, timestamp, and commit message.
* **Reading Order:** Read from **bottom (oldest)** to **top (newest)**.

### Compact One-Line Log (Recommended)
```bash
git log --oneline
```
* Displays each commit on a single, easy-to-read line with a **shortened 6–7 character commit ID** (enough to uniquely identify any commit), the **`HEAD -> master`** pointer, and the commit message.
* *Tip:* If you have many commits and the terminal pauses/hangs while viewing logs, press **`q`** to quit back to the prompt.

## 2. Ignoring Files & Folders (`.gitignore`)

In real projects, you often have sensitive files (e.g., `.env`, `API_secrets.txt`) or raw data folders (e.g., `bronze/`) that you **never** want to track or accidentally commit with `git add .`.

### How to Set Up `.gitignore`
1. Create a file named **`.gitignore`** in the **root (parent) directory** of your repository (at the same level as `.git`).
2. List the files and folders you want Git to ignore:

```text
# Ignore specific files
.env
API_secrets.txt

# Ignore an entire folder (include the trailing slash)
bronze/
```

3. Save the file, then stage and commit `.gitignore` itself:

```bash
git add .
git commit -m "add gitignore file"
```

* **Why track `.gitignore`?** It only contains the *names* of ignored files/folders—not the sensitive data inside them—so committing it ensures Git (and your teammates) consistently ignore those files.
* **Verify:** Run `git status`—the ignored files and folders will no longer appear as untracked.

## 3. Tracking Empty Folders (`.gitkeep`)

* **Important Git Rule:** Git **does not track empty folders**. If you create an empty folder (e.g., `silver/`) and run `git status`, Git will not detect it until you place a file inside it.
* **Why keep an empty folder?** Sometimes you want to preserve a project's folder structure (such as a `silver/` data layer) even before files are added to it.

### Workaround: `.gitkeep`
1. Create a hidden file named **`.gitkeep`** inside the empty folder (e.g., `silver/.gitkeep`).
2. Stage and commit the change:

```bash
git add .
git commit -m "add silver folder"
```

* Now Git tracks the folder and preserves its structure in the repository while keeping it effectively empty.
